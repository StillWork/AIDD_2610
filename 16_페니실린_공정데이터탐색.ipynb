## 🏭 의약품 제조 핵심 개념과 EDA의 관계

### QbD (Quality by Design) — 설계에 의한 품질
**QbD**는 ICH Q8 가이드라인에서 정의된 철학으로, "최종 검사로 품질을 확인"하는 것이 아니라 **"처음부터 공정을 이해하고 설계하여 품질을 내재화"**하는 접근 방식입니다.

- **CQA (Critical Quality Attribute)**: 환자 안전에 직접 영향 → 이 실습의 **페니실린 농도(g/L)** 가 대표적 CQA
- **CPP (Critical Process Parameter)**: CQA에 영향을 미치는 공정 변수 → 온도(T), pH, 용존산소(DO2), 기류율(Fg) 등
- **Design Space**: CPP들의 조합 범위 중 허용 가능한 공간. EDA → 상관관계 분석 → DOE → Design Space 순서로 도출

```
[발효조]
  CPP: 온도(298K), pH(6.5), DO2(mg/L), 기류율(L/h), 당 공급속도(L/h)
            ↓  (공정 제어)
  CQA: 페니실린 농도(g/L) ← 배치마다 다름 → EDA로 원인 파악
```

### PAT (Process Analytical Technology) — 공정분석기술
2004년 FDA 가이던스: 제조 공정을 **실시간으로 측정·분석·제어**하여 최종 품질을 보장하는 기술 체계

- 전통 방식: 생산 완료 후 샘플 채취 → 실험실 분석 → 합격/불합격 판정 (수 일 소요)
- PAT 방식: 공정 중 센서로 CPP 실시간 측정 → 즉시 예측 모델 실행 → 실시간 공정 제어 (수 분 내 피드백)
- **EDA는 PAT 구축의 첫 단계**: 어떤 CPP를 실시간 모니터링해야 하는지 결정하는 근거를 제공

### GMP (Good Manufacturing Practice) — 우수 의약품 제조관리기준
대한민국 식약처(MFDS), 미국 FDA, 유럽 EMA 모두 의약품 제조사에 GMP 준수를 요구합니다.

- **배치 기록(Batch Record)**: GMP에서는 배치 단위로 모든 공정 데이터를 완전히 기록·보관해야 함
- **연속 공정 검증(CPV, Continued Process Verification)**: 출시된 배치들의 공정 데이터를 지속적으로 분석하여 공정 상태 모니터링 → EDA가 CPV 분석의 핵심 도구
- **배치 기록의 AI 분석** = 이 실습의 목적

### 💡 이 실습이 실제 제약 현장에서 하는 일
항생제 공장에서는 이 노트북과 동일한 분석을 매 배치마다 수행합니다:
1. 발효조 센서 데이터(CPP) 수집 → 2. EDA로 배치 품질(CQA) 관계 파악 → 3. 불량 배치 원인 규명 → 4. 공정 파라미터 조정

## 💻 [셀 설명] 상관관계 히트맵 — QbD Risk Assessment

**이 셀이 하는 일:** 공정변수(CPP)들 간의 상관계수를 계산하고 색상으로 표현된 히트맵을 그립니다.

### 상관계수란?
두 변수가 함께 변화하는 **정도와 방향**을 -1 ~ +1 사이의 숫자로 표현합니다:
- **+0.7 이상** (진한 빨강): 강한 양의 상관 → 함께 증가/감소
- **-0.7 이하** (진한 파랑): 강한 음의 상관 → 한쪽 증가 시 반대쪽 감소
- **-0.3 ~ +0.3** (흰색): 상관 관계 거의 없음

### QbD에서 상관관계 분석의 역할
ICH Q8(R2)의 **Risk Assessment** 과정에서 필수 단계입니다:

```
[QbD 개발 흐름]
QTPP 정의 → CQA 결정 → Risk Assessment(상관관계 분석) → DOE → Design Space → Control Strategy
           (이 실습)        (이 셀!)
```

- **CQA와 높은 상관관계를 가진 CPP** = 가장 중요하게 관리해야 할 변수
- **CPP 간 높은 상관관계** = 다중공선성(Multicollinearity) 주의 → 회귀 모델에서 특성 선택 시 고려 필요
- **이 분석 결과** → 다음 실습(PM_02)에서 모델에 입력할 특성(Feature) 선택의 근거

## 💻 [셀 설명] 시계열 데이터 시각화 — PAT 모니터링 화면

**이 셀이 하는 일:** 처음 10,000개 샘플(약 1~2개 배치)의 모든 변수를 시간 흐름에 따라 그립니다.

### 왜 이렇게 많은 변수를 한꺼번에 보는가?
PAT(공정분석기술)의 핵심은 **공정 전체의 흐름을 파악**하는 것입니다. 발효 공정에서 변수들은 서로 연결되어 있습니다:

```
당 공급(Fs) ↑ → 기질 농도(S) ↑ → 균 성장 ↑ → 산소 소비 ↑ → DO2 ↓ → 기류율(Fg) ↑ 자동 조절
                                                              ↓
                                               CO2 발생 ↑ → 배기가스 CO2% ↑
```

### 이 시각화로 확인할 항목 (GMP 배치 기록 검토 시)
1. **트렌드(Trend)**: 시간에 따른 증가/감소 패턴이 정상인가?
2. **스파이크(Spike)**: 갑작스러운 이상값(Outlier)이 있는가?
3. **드리프트(Drift)**: 서서히 기준값에서 벗어나는 현상이 있는가?
4. **공정 단계**: 전반부(균 성장기)와 후반부(페니실린 생산기)의 차이가 보이는가?

> **GMP 실무:** 이와 같은 시각화는 **배치 분석 보고서(Batch Analysis Report)**에 포함되며, QA(Quality Assurance) 부서의 배치 심사(Batch Record Review)에 활용됩니다.

## 공정변수 상관관계 분석

````python
# 연속형 공정변수 선택 및 상관관계 히트맵
numeric_cols = [
    'Temperature(T:K)', 'pH(pH:pH)',
    'Dissolved oxygen concentration(DO2:mg/L)',
    'Aeration rate(Fg:L/h)', 'Sugar feed rate(Fs:L/h)',
    'Penicillin concentration(P:g/L)',
    'Substrate concentration(S:g/L)',
    'carbon dioxide percent in off-gas(CO2outgas:%)',
    'Oxygen in percent in off-gas(O2:O2  (%))',
]
numeric_cols = [c for c in numeric_cols if c in df.columns]

# 배치 1 데이터로 상관관계 계산
corr = df[df['Batch ID'] == 1][numeric_cols].corr()

short_names = ['온도', 'pH', '용존산소', '통기율', '당공급률',
               '페니실린농도', '기질농도', 'CO2(배기)', 'O2(배기)']
corr.index = short_names[:len(corr)]
corr.columns = short_names[:len(corr)]

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r',
            center=0, vmin=-1, vmax=1, ax=ax,
            annot_kws={'size': 9})
ax.set_title('공정변수 상관관계 히트맵 (배치 #1)', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
print('💡 페니실린 농도와 상관관계가 높은 변수 파악 → 다음 단계 회귀 모델의 특성 선택에 활용')
````
