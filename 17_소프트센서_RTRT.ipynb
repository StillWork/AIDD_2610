## 🧫 페니실린 발효 4단계 — 왜 초기 30시간 예측이 어려운가?

```
시간(h):  0      20      60                    180       227
          │지연기 │  성장기  │     생산기(이디오파즈) ★    │사멸│
P(g/L):   ≈0 ──────── ≈0 ──┐
                             └──────────────────────────── Peak
```

- **지연기·성장기 (0~60h)**: 균이 성장하는 단계, 페니실린은 거의 생산되지 않음
- **생산기 (60~180h)**: 성장이 느려지면서 **이차대사산물**인 페니실린이 급격히 축적됨
- **초기 30h의 CPP** 패턴은 최종 수율과 상관이 낮음 → 직접 수율 예측 R² ≈ -1

**해결책**: 시간 이동(time shift) 없이 *같은 시점* CPP → P를 예측하면  
물리화학적 상관관계가 직접 활용되어 R² > 0.95를 달성할 수 있습니다.

# 모델 A — 소프트 센서 (Soft Sensor)
## 🎯 시나리오: CPP 8개로 페니실린 농도를 실시간 추정

페니실린 발효조에서 P:g/L(페니실린 농도)를 직접 측정하려면  
**오프라인 샘플링 → HPLC 분석 → 결과 반환까지 수 시간**이 걸립니다.  

소프트 센서는 **이미 실시간으로 측정되는 CPP**를 이용해 P를 즉각 추정합니다.

```
[실시간 온라인 센서]          [소프트 센서 AI]      [PAT 대시보드]
 T=298K, pH=6.5,         →   RF 모델           →  P̂ = 2.31 g/L
 DO₂=4.2, Fg=65,              (학습된 매핑)          ± 0.05 g/L
 Fs=80, S=0.3, …
```

| 구분 | 오프라인 측정 | 소프트 센서 |
|------|------------|-----------|
| 결과 지연 | 2~6시간 | 수 초 이내 |
| 비용 | 높음 (시약·인력) | 낮음 (모델 추론) |
| 연속성 | 간헐적 | 실시간 연속 |
| FDA 근거 | 필수 분석 | PAT Guidance 2004 허용 |

## 💻 [셀 설명] 소프트 센서 데이터 준비 — 배치 기준 분리

**주의**: 행(row) 단위 랜덤 분리는 **같은 배치의 앞·뒤 시점이 훈련/검증에 섞이는 데이터 누수**를 일으킵니다.  
배치 번호로 분리해야 독립적인 성능 평가가 가능합니다.

- 훈련: 정상 배치 1~70 (약 80,000 샘플)
- 검증: 정상 배치 71~90 (약 23,000 샘플)
- 이상 배치(91~100)는 별도 평가용으로 보존

````python
# 정상 배치만 사용
df_normal = df_raw[df_raw[FAULT_COL] == 0].copy()

# 배치 기준 분리
tr_rows = df_normal[df_normal[BATCH_COL] <= 70][CPP_COLS + [P_COL]].dropna()
te_rows = df_normal[df_normal[BATCH_COL] >  70][CPP_COLS + [P_COL]].dropna()

X_tr_a = tr_rows[CPP_COLS].values
y_tr_a = tr_rows[P_COL].values
X_te_a = te_rows[CPP_COLS].values
y_te_a = te_rows[P_COL].values

print(f"훈련 샘플: {len(X_tr_a):,}  |  검증 샘플: {len(X_te_a):,}")
print(f"P:g/L 훈련 범위: {y_tr_a.min():.2f} ~ {y_tr_a.max():.2f}")
````

## 💻 [셀 설명] Random Forest 소프트 센서 모델 훈련

각 행(row) = 한 시점의 CPP 8개 측정값 → 같은 시점의 P:g/L  
시간 이동 없이 **동시점 매핑**을 학습하므로 물리화학적 상관관계를 직접 포착합니다.

````python
rf_ss = RandomForestRegressor(
    n_estimators=200, max_depth=15,
    min_samples_leaf=5, random_state=42, n_jobs=-1
)
rf_ss.fit(X_tr_a, y_tr_a)

y_pred_a = rf_ss.predict(X_te_a)

mae_a  = mean_absolute_error(y_te_a, y_pred_a)
mse_a = mean_squared_error(y_te_a, y_pred_a)
rmse_a = np.sqrt(mse_a)
r2_a   = r2_score(y_te_a, y_pred_a)

print("=" * 50)
print("  모델 A — 소프트 센서 성능 (검증 배치 71~90)")
print(f"  MAE  = {mae_a:.4f} g/L")
print(f"  RMSE = {rmse_a:.4f} g/L")
print(f"  R²   = {r2_a:.4f}  ({r2_a*100:.1f}%)")
print("=" * 50)
````

## 💻 [셀 설명] 소프트 센서 결과 시각화

**왼쪽 산점도**: 실제 P 측정값 vs AI 추정값 (대각선에 가까울수록 정확)  
**오른쪽 히스토그램**: 예측 오차 분포 — 정규분포에 가까울수록 모델이 편향 없이 균일하게 예측함

````python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# ① 실제 vs 예측 산점도 (5,000개 샘플링)
idx = np.random.choice(len(y_te_a), min(5000, len(y_te_a)), replace=False)
lim = (min(y_te_a[idx].min(), y_pred_a[idx].min()) * 0.95,
       max(y_te_a[idx].max(), y_pred_a[idx].max()) * 1.05)

axes[0].scatter(y_te_a[idx], y_pred_a[idx], s=4, alpha=0.25, color='royalblue')
axes[0].plot(lim, lim, 'r--', linewidth=1.5, label='완벽 예측 (y=x)')
axes[0].set_xlim(lim); axes[0].set_ylim(lim)
axes[0].set_xlabel('실제 페니실린 농도 (g/L)')
axes[0].set_ylabel('소프트 센서 예측값 (g/L)')
axes[0].set_title(f'소프트 센서 성능  R²={r2_a:.4f}')
axes[0].legend()
axes[0].text(0.05, 0.92, f'MAE={mae_a:.4f}  RMSE={rmse_a:.4f}',
             transform=axes[0].transAxes, fontsize=9, color='navy')

# ② 오차 분포
errors = y_pred_a - y_te_a
axes[1].hist(errors, bins=60, color='steelblue', alpha=0.7, edgecolor='white')
axes[1].axvline(0, color='red', linestyle='--', linewidth=1.5, label='오차=0')
axes[1].axvline(errors.mean(), color='orange', linestyle='-', linewidth=1.5,
                label=f'평균 오차: {errors.mean():.4f}')
axes[1].set_xlabel('예측 오차 (g/L)')
axes[1].set_ylabel('샘플 수')
axes[1].set_title('소프트 센서 예측 오차 분포')
axes[1].legend()

plt.tight_layout()
plt.show()
````

## 💻 [셀 설명] 단일 배치 시계열 추적 — "실시간 P 모니터링" 시뮬레이션

배치 하나를 골라 **배치 전 구간에서 소프트 센서가 실시간으로 P를 추정**하는 모습을 시각화합니다.  
이 그래프가 PAT 대시보드에 표시될 화면과 동일한 개념입니다.

````python
VIZ_BATCH_A = 75  # 시각화 배치 (71~90 중 선택)

bdf_viz = df_raw[df_raw[BATCH_COL] == VIZ_BATCH_A].sort_values(TIME_COL) if TIME_COL in df_raw.columns           else df_raw[df_raw[BATCH_COL] == VIZ_BATCH_A].reset_index(drop=True)

valid_mask = bdf_viz[CPP_COLS + [P_COL]].notna().all(axis=1)
bdf_viz    = bdf_viz[valid_mask].reset_index(drop=True)

X_viz = bdf_viz[CPP_COLS].values
y_viz = bdf_viz[P_COL].values
y_viz_pred = rf_ss.predict(X_viz)

time_axis = bdf_viz[TIME_COL].values if TIME_COL in bdf_viz.columns else np.arange(len(bdf_viz)) * SAMPLE_INTERVAL

plt.figure(figsize=(12, 5))
plt.plot(time_axis, y_viz,      color='royalblue', linewidth=2, label='실제 P:g/L (온라인 측정)')
plt.plot(time_axis, y_viz_pred, color='darkorange', linewidth=1.5,
         linestyle='--', label='소프트 센서 추정값', alpha=0.9)

# 오프라인 측정 지점 (간헐적 샘플링 예시)
offline_idx = np.arange(0, len(time_axis), int(6/SAMPLE_INTERVAL))  # 6h마다 오프라인 측정 가정
plt.scatter(time_axis[offline_idx], y_viz[offline_idx],
            color='green', s=30, zorder=5, label='오프라인 분석 시점 (6h 간격 예시)')

plt.xlabel('배치 경과 시간 (h)')
plt.ylabel('페니실린 농도 (g/L)')
plt.title(f'배치 {VIZ_BATCH_A} — PAT 소프트 센서 실시간 추적 시뮬레이션')
plt.legend(fontsize=9)
plt.tight_layout()
plt.show()

ss_mae = mean_absolute_error(y_viz, y_viz_pred)
ss_r2  = r2_score(y_viz, y_viz_pred)
print(f"배치 {VIZ_BATCH_A}  MAE={ss_mae:.4f} g/L  R²={ss_r2:.4f}")
````

# 모델 C — 배치 진행률별 최종 수율 예측 (RTRT 시점 분석)
## 🎯 시나리오: "언제부터 배치 결과를 신뢰할 수 있나?"

배치가 X% 진행된 시점까지의 공정 데이터로 **배치 종료 시 최대 P:g/L**을 예측합니다.  
X를 30%~90%로 변화시키며 R² 변화를 추적하면  
**"어느 시점부터 예측이 신뢰할 수 있는가?"**를 정량적으로 파악할 수 있습니다.

```
배치 진행률  예측 신뢰도(R²)  GMP 의미
───────────────────────────────────────
   30 %         낮음          생산기 미진입, 예측 불가
   50 %         중간          P 상승 시작, 일부 예측 가능
   70 %         높음          P 곡선 확립, RTRT 적용 고려
   90 %         매우 높음      거의 확정적
```

## 💻 [셀 설명] 배치 진행률별 특성 추출 및 RF 반복 학습

각 임계값(threshold)에 대해:
1. 배치별로 "전체 길이 × threshold" 만큼의 앞부분만 사용
2. CPP별 통계(mean·std·min·max·slope) 추출 → 특성 벡터 1개/배치
3. 목표 = 해당 배치의 최대 P:g/L (최종 수율)
4. Random Forest 회귀 → 성능 기록

**Slope(선형 기울기)** 를 추가하는 이유: 현재 수준뿐 아니라 **트렌드 방향**이 최종 수율 예측에 중요합니다.

````python
THRESHOLDS = [0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]
TRAIN_CUTOFF = 72  # 배치 1~72: 훈련, 73~90: 검증

results_c = []

for thr in THRESHOLDS:
    feature_rows = []

    for batch_id, bdf in df_raw[df_raw[FAULT_COL] == 0].groupby(BATCH_COL):
        bdf = bdf.sort_values(TIME_COL).reset_index(drop=True) if TIME_COL in df_raw.columns               else bdf.reset_index(drop=True)
        n_total = len(bdf)
        cutoff  = max(int(n_total * thr), 10)
        early   = bdf.iloc[:cutoff]

        row = {BATCH_COL: batch_id,
               'max_P': bdf[P_COL].max()}

        for col in CPP_COLS:
            vals = early[col].dropna().values
            if len(vals) < 3:
                row.update({f'{col}_mean':0, f'{col}_std':0,
                            f'{col}_min':0,  f'{col}_max':0, f'{col}_slope':0})
                continue
            row[f'{col}_mean']  = vals.mean()
            row[f'{col}_std']   = vals.std(ddof=0)
            row[f'{col}_min']   = vals.min()
            row[f'{col}_max']   = vals.max()
            # 선형 기울기 (트렌드)
            row[f'{col}_slope'] = np.polyfit(np.arange(len(vals)), vals, 1)[0]

        feature_rows.append(row)

    feat = pd.DataFrame(feature_rows).fillna(0)
    FEAT_COLS_C = [c for c in feat.columns if c not in [BATCH_COL, 'max_P']]

    tr = feat[feat[BATCH_COL] <= TRAIN_CUTOFF]
    te = feat[feat[BATCH_COL] >  TRAIN_CUTOFF]

    rf_c = RandomForestRegressor(n_estimators=200, max_depth=10,
                                  min_samples_leaf=2, random_state=42, n_jobs=-1)
    rf_c.fit(tr[FEAT_COLS_C], tr['max_P'])

    y_true_c = te['max_P'].values
    y_pred_c = rf_c.predict(te[FEAT_COLS_C])

    mse_c = mean_squared_error(y_true_c, y_pred_c)
    rmse_c = np.sqrt(mse_c)

    results_c.append({
        'threshold': thr,
        'pct': int(thr * 100),
        'r2':   r2_score(y_true_c, y_pred_c),
        'mae':  mean_absolute_error(y_true_c, y_pred_c),
        'rmse': rmse_c,
        'y_true': y_true_c,
        'y_pred': y_pred_c,
        'batch_ids': te[BATCH_COL].values,
        'model':  rf_c,
        'feat_cols': FEAT_COLS_C,
        'feat_df': feat,
    })
    print(f"  {int(thr*100):3d}%  →  R²={results_c[-1]['r2']:+.4f}  MAE={results_c[-1]['mae']:.4f} g/L")
````

## 💻 [셀 설명] R² 변화 곡선 — RTRT 적용 가능 시점 식별

이 그래프가 핵심입니다.  
- x축: 배치 진행률 (%)  
- y축: 최종 수율 예측 R²  

R²가 **실용 임계값(예: 0.80)을 넘는 첫 시점**이 RTRT 판단의 과학적 근거가 됩니다.

````python
df_res = pd.DataFrame([{k:v for k,v in r.items()
                         if k not in ('y_true','y_pred','model','feat_cols','feat_df','batch_ids')}
                        for r in results_c])

RTRT_THRESHOLD = 0.80  # RTRT 적용 고려 기준선

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ① R² 변화 곡선
ax = axes[0]
ax.plot(df_res['pct'], df_res['r2'], 'o-', color='royalblue', linewidth=2.5,
        markersize=8, label='검증 R²')
ax.axhline(y=RTRT_THRESHOLD, color='orange', linestyle='--', linewidth=1.5,
           label=f'RTRT 기준선 R²={RTRT_THRESHOLD}')
ax.axhline(y=0, color='red', linestyle=':', linewidth=1, label='R²=0 (무의미)')

# 기준선 넘는 첫 시점 표시
for r in results_c:
    if r['r2'] >= RTRT_THRESHOLD:
        ax.axvline(x=r['pct'], color='green', linestyle='-.', linewidth=1.5,
                   label=f"RTRT 가능 시점: {r['pct']}%", alpha=0.7)
        ax.annotate(f"{r['pct']}%\nR²={r['r2']:.2f}",
                    xy=(r['pct'], r['r2']), xytext=(r['pct']+3, r['r2']-0.08),
                    fontsize=9, color='darkgreen',
                    arrowprops=dict(arrowstyle='->', color='darkgreen'))
        break

ax.set_xlim(25, 95); ax.set_ylim(-0.3, 1.05)
ax.set_xlabel('배치 진행률 (%)')
ax.set_ylabel('예측 R²')
ax.set_title('배치 진행률별 최종 수율 예측 R²')
ax.legend(fontsize=9); ax.grid(alpha=0.3)

# ② MAE 변화 곡선
ax2 = axes[1]
ax2.plot(df_res['pct'], df_res['mae'],  's-', color='tomato',    linewidth=2.5,
         markersize=7, label='MAE (g/L)')
ax2.plot(df_res['pct'], df_res['rmse'], '^-', color='darkorange', linewidth=2,
         markersize=7, label='RMSE (g/L)')
ax2.set_xlabel('배치 진행률 (%)')
ax2.set_ylabel('예측 오차 (g/L)')
ax2.set_title('배치 진행률별 수율 예측 오차')
ax2.legend(); ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n진행률별 성능 요약:")
print(f"{'진행률':>6}  {'R²':>8}  {'MAE':>8}  {'RMSE':>8}")
print("-" * 40)
for r in results_c:
    flag = " ← RTRT 가능" if r['r2'] >= RTRT_THRESHOLD else ""
    print(f"  {r['pct']:3d}%   {r['r2']:+.4f}   {r['mae']:.4f}   {r['rmse']:.4f}{flag}")
````

## 💻 [셀 설명] RTRT 가능 시점의 예측 결과 상세 확인

R²가 기준을 처음 넘는 시점(RTRT 가능 시점)의  
실제 vs 예측 산점도를 그립니다.  
이 산점도가 FDA 제출용 소프트웨어 검증 보고서(SVR)의 핵심 그래프입니다.

````python
# RTRT 기준 이상인 첫 번째 threshold 선택
rtrt_result = next((r for r in results_c if r['r2'] >= RTRT_THRESHOLD), results_c[-1])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# ① 산점도
y_true_r = rtrt_result['y_true']
y_pred_r = rtrt_result['y_pred']
pct_r    = rtrt_result['pct']

lim = (min(y_true_r.min(), y_pred_r.min()) * 0.92,
       max(y_true_r.max(), y_pred_r.max()) * 1.08)

axes[0].scatter(y_true_r, y_pred_r, color='royalblue', s=80, zorder=3, alpha=0.8)
axes[0].plot(lim, lim, 'r--', linewidth=1.5)

# 배치 번호 표시
for bid, yt, yp in zip(rtrt_result['batch_ids'], y_true_r, y_pred_r):
    axes[0].annotate(str(bid), (yt, yp), textcoords='offset points',
                     xytext=(4, 2), fontsize=7, color='gray')

axes[0].set_xlim(lim); axes[0].set_ylim(lim)
axes[0].set_xlabel('실제 최종 수율 (g/L)')
axes[0].set_ylabel('예측 최종 수율 (g/L)')
axes[0].set_title(f'''배치 {pct_r}% 시점 → 최종 수율 예측
                  R²={rtrt_result["r2"]:.4f}  MAE={rtrt_result["mae"]:.4f} g/L''')

# ② Feature Importance (상위 12개)
imp   = rtrt_result['model'].feature_importances_
fcols = rtrt_result['feat_cols']
top_n = 12
top_idx = np.argsort(imp)[-top_n:]

short = []
for n in [fcols[i] for i in top_idx]:
    for full, short_name in DISPLAY.items():
        n = n.replace(full, short_name)
    short.append(n.replace('_mean','_평균').replace('_std','_표준편차')
                  .replace('_min','_최소').replace('_max','_최대').replace('_slope','_기울기'))

colors_imp = ['#1565C0' if v >= sorted(imp)[-3] else '#90CAF9' for v in imp[top_idx]]
axes[1].barh(short, imp[top_idx], color=colors_imp)
axes[1].set_xlabel('Feature Importance')
axes[1].set_title(f'수율 예측 핵심 특성 Top {top_n} (배치 {pct_r}% 기준)')
plt.tight_layout()
plt.show()
````

## 💻 [셀 설명] 실시간 배치 모니터링 시뮬레이션

배치가 진행되면서 RTRT 예측값이 어떻게 수렴하는지 보여줍니다.  
각 진행률 시점에서의 예측값과 실제 최종 수율을 비교합니다.

````python
# 특정 배치(예: 배치 80)의 수율을 각 진행률 시점에서 예측
TARGET_BATCH = 80

actuals, preds, pcts = [], [], []
for r in results_c:
    mask = r['batch_ids'] == TARGET_BATCH
    if mask.any():
        actuals.append(r['y_true'][mask][0])
        preds.append(r['y_pred'][mask][0])
        pcts.append(r['pct'])

if pcts:
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.plot(pcts, preds,   'o-', color='darkorange', linewidth=2.5,
            markersize=9, label='AI 예측 수율 (g/L)', zorder=3)
    ax.axhline(y=actuals[0], color='royalblue', linestyle='--', linewidth=2,
               label=f'실제 최종 수율: {actuals[0]:.2f} g/L')

    # 오차 범위 (±MAE)
    for r in results_c:
        mask = r['batch_ids'] == TARGET_BATCH
        if mask.any():
            yp = r['y_pred'][mask][0]
            ax.errorbar(r['pct'], yp, yerr=r['mae'], fmt='none',
                        ecolor='orange', elinewidth=1.5, capsize=4, alpha=0.6)

    # RTRT 가능 구간 표시
    rtrt_pct = rtrt_result['pct']
    ax.axvspan(rtrt_pct, 100, alpha=0.07, color='green', label=f'RTRT 적용 가능 구간 (≥{rtrt_pct}%)')
    ax.axvline(x=rtrt_pct, color='green', linestyle='-.', linewidth=1.5)

    ax.set_xlabel('배치 진행률 (%)')
    ax.set_ylabel('페니실린 최종 수율 예측 (g/L)')
    ax.set_title(f'배치 {TARGET_BATCH} — 진행률별 수율 예측 수렴 과정 (오차 막대 = MAE)')
    ax.legend(fontsize=9)
    ax.set_xlim(25, 95)
    plt.tight_layout()
    plt.show()
else:
    print(f"배치 {TARGET_BATCH}는 검증 셋에 없습니다. TARGET_BATCH를 73~90 사이로 변경해 주세요.")
````

## 📝 모델 C — GMP RTRT 대응 해석

### RTRT 의사결정 흐름 (FDA PAT Guidance 2004 기반)

```
배치 시작
    │
    ├─ 진행률 30~50% : AI 예측 낮은 신뢰도 → 공정 모니터링 집중
    │
    ├─ 진행률 [RTRT 가능 시점]% : R² ≥ 0.80 → 예측 수율이 기준치 초과?
    │         ┌── YES → 공정 파라미터 미세 조정 또는 조기 수확 준비
    │         └── NO  → 보정 조치 (당 공급량 증가, 기류율 조정)
    │
    └─ 배치 종료 → 최종 샘플링 · 출하 검토 (소프트 센서 예측과 비교 검증)
```

| R² 단계 | 상태 | 조치 |
|---------|------|------|
| < 0.5 (초기) | 예측 불가 | 공정 모니터링만 수행 |
| 0.5~0.8 | 참고 수준 | 보조 지표로만 활용 |
| ≥ 0.8 (RTRT 가능) | 신뢰 예측 | RTRT 판단 근거로 활용 가능 |
| ≥ 0.95 | 고신뢰 | 오프라인 분석 감소 검토 가능 |

> **ICH Q10 Continual Improvement**: 배치가 누적될수록 훈련 데이터가 늘어나  
> 모델 R²가 점진적으로 향상됩니다. 이를 **지속적 공정 검증(CPV)**의 AI 구현으로 볼 수 있습니다.

# 모델 B — 20시간 선행 예측 (PAT Soft Sensor — 미래 예측)

> 현재 공정 상태로 **20시간 후 페니실린 농도를 예측**합니다.  
> 모델 A(현재 예측)와 달리, 20h 앞을 내다보는 **예측형 PAT** 입니다.

## 💻 [셀 설명] Sliding Window Feature Engineering (20h 선행)

```
과거 10h 윈도우                     예측 목표
[t-50 ··· t]  →  통계 특성 벡터  →  P at (t + 100스텝 ≈ t+20h)
 mean/std/last/slope × 9개 변수
```

````python
WINDOW_H  = 10   # 입력 윈도우 (h)
HORIZON_H = 20   # 예측 선행 시간 (h)

WINDOW_STEPS  = max(int(WINDOW_H  / SAMPLE_INTERVAL), 5)
HORIZON_STEPS = max(int(HORIZON_H / SAMPLE_INTERVAL), 10)
FEAT_COLS_B   = CPP_COLS + [P_COL]

def window_feats(wdf):
    f = []
    for col in FEAT_COLS_B:
        v = wdf[col].dropna().values
        if len(v) < 2: f += [0,0,0,0]; continue
        f.append(v.mean())
        f.append(v.std(ddof=0))
        f.append(v[-1])
        f.append(np.polyfit(np.arange(len(v)), v, 1)[0])
    return f

X_tr_b, y_tr_b, X_te_b, y_te_b = [], [], [], []

for bid, bdf in df_raw[df_raw[FAULT_COL] == 0].groupby(BATCH_COL):
    bdf = (bdf.sort_values(TIME_COL) if TIME_COL in bdf.columns else bdf).reset_index(drop=True)
    n = len(bdf)
    for i in range(WINDOW_STEPS, n - HORIZON_STEPS):
        wf = window_feats(bdf.iloc[i - WINDOW_STEPS:i])
        tg = bdf.iloc[i + HORIZON_STEPS][P_COL]
        if np.isnan(tg): continue
        (X_tr_b if bid <= 70 else X_te_b).append(wf)
        (y_tr_b if bid <= 70 else y_te_b).append(tg)

X_tr_b = np.array(X_tr_b); y_tr_b = np.array(y_tr_b)
X_te_b = np.array(X_te_b); y_te_b = np.array(y_te_b)
print(f"훈련: {len(X_tr_b):,}  검증: {len(X_te_b):,}  특성: {X_tr_b.shape[1]}")
````

````python
rf_b = RandomForestRegressor(n_estimators=50, max_depth=12,
                             min_samples_leaf=5, random_state=42, n_jobs=-1)
rf_b.fit(X_tr_b, y_tr_b)
y_pred_b = rf_b.predict(X_te_b)

mae_b  = mean_absolute_error(y_te_b, y_pred_b)
mse_b = mean_squared_error(y_te_b, y_pred_b)
rmse_b = np.sqrt(mse_b)
r2_b   = r2_score(y_te_b, y_pred_b)

print(f"모델 B — {HORIZON_H}h 선행 예측")
print(f"  MAE={mae_b:.4f}  RMSE={rmse_b:.4f}  R²={r2_b:.4f}")
````

````python
# 단일 배치 롤링 예측 시각화
VIZ_B = 75
bdf_b = df_raw[df_raw[BATCH_COL] == VIZ_B]
bdf_b = (bdf_b.sort_values(TIME_COL) if TIME_COL in bdf_b.columns else bdf_b).reset_index(drop=True)
bdf_b = bdf_b[FEAT_COLS_B].reset_index(drop=True)

p_steps, p_vals = [], []
for i in range(WINDOW_STEPS, len(bdf_b) - HORIZON_STEPS):
    p_steps.append(i + HORIZON_STEPS)
    p_vals.append(rf_b.predict([window_feats(bdf_b.iloc[i-WINDOW_STEPS:i])])[0])

t_full = (df_raw[df_raw[BATCH_COL]==VIZ_B].sort_values(TIME_COL)[TIME_COL].values
          if TIME_COL in df_raw.columns else np.arange(len(bdf_b)) * SAMPLE_INTERVAL)
t_pred = [t_full[s] if s < len(t_full) else t_full[-1] for s in p_steps]

plt.figure(figsize=(12, 4))
plt.plot(t_full, bdf_b[P_COL].values, 'b-', linewidth=1.5, label='실제 P:g/L')
plt.plot(t_pred, p_vals, '--', color='darkorange', linewidth=1.5,
         label=f'AI 예측 (20h 선행)')
plt.xlabel('배치 경과 시간 (h)'); plt.ylabel('페니실린 농도 (g/L)')
plt.title(f'배치 {VIZ_B} — 20h 선행 예측 vs 실제')
plt.legend(fontsize=9); plt.tight_layout(); plt.show()
````
